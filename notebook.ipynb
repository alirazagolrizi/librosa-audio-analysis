# Audio Waveform & Spectrogram Analysis | تحلیل شکل موج و طیف‌نگار صوت

Load user-supplied audio with librosa, plot waveform/spectrogram, and print spectral features.

Open the repository README for prerequisites, input assets, limitations, and run instructions. | برای پیش‌نیازها، ورودی‌ها و راهنمای اجرا README مخزن را ببینید.

```python
import librosa
import librosa.display
import matplotlib.pyplot as plt

def show_wave_and_spectrogram(file_path, title_prefix):
    y, sr = librosa.load(file_path, sr=None)
    plt.figure(figsize=(12, 4))

    # شکل موج
    plt.subplot(1, 2, 1)
    librosa.display.waveshow(y, sr=sr)
    plt.title(f"{title_prefix} - Waveform")

    # اسپکتروگرام
    plt.subplot(1, 2, 2)
    S = librosa.stft(y)
    S_db = librosa.amplitude_to_db(abs(S))
    librosa.display.specshow(S_db, sr=sr, x_axis='time', y_axis='hz')
    plt.title(f"{title_prefix} - Spectrogram")
    plt.colorbar(format="%+2.0f dB")

    plt.tight_layout()
    plt.show()

    # چند ویژگی فرکانسی ساده
    centroid = librosa.feature.spectral_centroid(y=y, sr=sr).mean()
    bandwidth = librosa.feature.spectral_bandwidth(y=y, sr=sr).mean()
    print(f"{title_prefix} - Spectral Centroid:", centroid)
    print(f"{title_prefix} - Spectral Bandwidth:", bandwidth)
    print("-" * 60)

show_wave_and_spectrogram("calm.wav", "Calm Environment")
show_wave_and_spectrogram("fan.wav", "Fan Noise Environment")
```
